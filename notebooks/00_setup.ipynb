```python
%sql
CREATE CATALOG IF NOT EXISTS retail_lakehouse
COMMENT 'Databricks Retail Orders Lakehouse project';
```

```python
%sql
USE CATALOG retail_lakehouse;

CREATE SCHEMA IF NOT EXISTS bronze COMMENT 'Raw data as ingested, no changes';
CREATE SCHEMA IF NOT EXISTS silver COMMENT 'Cleaned, deduplicated, validated data';
CREATE SCHEMA IF NOT EXISTS gold   COMMENT 'Business-ready star schema for analytics and ML';
```

```python
%sql
USE CATALOG retail_lakehouse;

CREATE VOLUME IF NOT EXISTS bronze.raw_files
COMMENT 'Landing zone for raw CSV files';

CREATE VOLUME IF NOT EXISTS bronze.checkpoints
COMMENT 'Auto Loader streaming checkpoints and schema tracking';
```
