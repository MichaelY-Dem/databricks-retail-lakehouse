# 03 - Silver Layer Checks
Confirms row counts, no duplicates, bad rows removed, and late-arriving data handled correctly.


```python
%sql
SELECT 'orders' AS table_name, COUNT(*) AS row_count FROM retail_lakehouse.silver.orders
UNION ALL SELECT 'order_items', COUNT(*) FROM retail_lakehouse.silver.order_items
UNION ALL SELECT 'customers',   COUNT(*) FROM retail_lakehouse.silver.customers
UNION ALL SELECT 'products',    COUNT(*) FROM retail_lakehouse.silver.products
UNION ALL SELECT 'payments',    COUNT(*) FROM retail_lakehouse.silver.payments;
```

```python
%sql
SELECT
  (SELECT COUNT(*) FROM retail_lakehouse.silver.orders WHERE order_id IS NULL)            AS null_order_ids,
  (SELECT COUNT(*) FROM (SELECT order_id FROM retail_lakehouse.silver.orders
                         GROUP BY order_id HAVING COUNT(*) > 1))                           AS duplicate_order_ids;
```
